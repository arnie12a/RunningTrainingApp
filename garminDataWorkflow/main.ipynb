```python
import pandas as pd
```

```python
hike_df = pd.read_csv('./data/hiking.csv')
hike_df
```

```python
run_df = pd.read_csv('./data/runs.csv')
run_df
```

```python
# run_df["Time"] = run_df["Time"].astype(str).str.replace(r"\.\d+$", "", regex=True)
run_df.loc[:, "Time"] = (
    run_df.loc[:, "Time"]
    .astype(str)
    .str.replace(r"\.\d+$", "", regex=True)
)

run_df.loc[:, "Distance"] = (
    run_df["Distance"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

run_df.loc[:, "Distance"] = pd.to_numeric(run_df["Distance"], errors="coerce")
track_mask = run_df["Title"] == "Track Running"

run_df.loc[track_mask, "Distance"] = (
    run_df.loc[track_mask, "Distance"] * 0.000621371
)
run_df.loc[:, "Distance"] = run_df["Distance"].round(2)



```

```python
# Assuming hike_df and run_df are already defined DataFrames
# Convert hike_df to a list of dictionaries
hike_json = hike_df.to_json(orient='records')

# Convert run_df to a list of dictionaries
run_json = run_df.to_json(orient='records')

# Save the JSON strings to files
with open('hike_data.json', 'w') as hike_file:
    hike_file.write(hike_json)

with open('run_data.json', 'w') as run_file:
    run_file.write(run_json)
```
