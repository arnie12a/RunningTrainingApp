```python
run_cols = ['Date', 'Title', 'Distance', 'Time', 'Avg HR', 'Max HR', 'Avg Pace', 'Total Ascent', 'Total Descent', 'Moving Time', 'Elapsed Time', 'Min Elevation', 'Max Elevation', 'Body Battery Drain']
```

```python
hike_df = hikes[hike_cols]
hike_df
```

```python
run_df = runs[run_cols]
run_df
```

```python
# run_df["Time"] = run_df["Time"].astype(str).str.replace(r"\.\d+$", "", regex=True)
run_df.loc[:, "Time"] = (
    run_df.loc[:, "Time"]
    .astype(str)
    .str.replace(r"\.\d+$", "", regex=True)
)

# # Ensure Distance is numeric
# run_df.loc[:, "Distance"] = pd.to_numeric(run_df["Distance"], errors="coerce")

# # Convert Track Running distances from meters to miles
# track_mask = run_df["Title"] == "Track Running"

# run_df.loc[track_mask, "Distance"] = (
#     run_df.loc[track_mask, "Distance"] * 0.000621371
# )

# # Optional: round for readability
# run_df.loc[:, "Distance"] = run_df["Distance"].round(2)
run_df.loc[:, "Distance"] = (
    run_df["Distance"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

run_df.loc[:, "Distance"] = pd.to_numeric(run_df["Distance"], errors="coerce")
track_mask = run_df["Title"] == "Track Running"

run_df.loc[track_mask, "Distance"] = (
    run_df.loc[track_mask, "Distance"] * 0.000621371
)
run_df.loc[:, "Distance"] = run_df["Distance"].round(2)



```

```python
run_df
```

```python
hike_df = hike_df.rename(columns={'Ascent': 'Total Ascent'})
hike_df
```

```python
result = pd.concat([hike_df, run_df], ignore_index=True)
result
```

```python
import pandas as pd
import json
import numpy as np

# Assuming your dataframe is called df
df = result.copy()

# Replace NaN with None for JSON
df = df.where(pd.notnull(df), None)

# Function to infer TerrainType
def infer_terrain(title):
    title = str(title).lower()
    if "treadmill" in title:
        return "Treadmill"
    if "track" in title:
        return "Track"
    if "hiking" in title:
        return "Trail"
    return "Road"

# Function to infer RunType
def infer_run_type(title):
    title = str(title).lower()
    if "hiking" in title:
        return "Hike"
    return "Run"

# -----------------------------
# Format Date
# -----------------------------
run_df["Date"] = pd.to_datetime(run_df["Date"]).dt.strftime("%Y-%m-%d")

# Build JSON list dynamically
json_data = []

for _, row in df.iterrows():
    json_data.append({
        "Date": row.get("Date"),
        "Distance": row.get("Distance"),
        "TotalTime": row.get("Time"),
        "MovingTime": row.get("Moving Time"),
        "AveragePace": row.get("Avg Pace"),
        "ElevationGain": row.get("Total Ascent"),
        "ElevationDescent": row.get("Total Descent"),
        "TerrainType": infer_terrain(row.get("Title")),
        "RunType": infer_run_type(row.get("Title")),
        "AvgHeartRate": row.get("Avg HR"),
        "MaxHeartRate": row.get("Max HR"),
        "BodyBatteryDrain": row.get("Body Battery Drain")
    })

# Write to JSON
with open("ultra50km.json", "w") as f:
    json.dump(json_data, f, indent=2)

print("✅ ultra50km.json created successfully!")
```
